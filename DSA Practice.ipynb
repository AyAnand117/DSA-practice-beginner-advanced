#### Extraction of Digits 

```python
n = 5873
# 5873 % 10 = 3
# 5873 // 10 = 587
# 587 % 10 = 7
# 587 // 10 = 58
# 58 % 10 = 8
# 58 // 10 = 5
# 5 % 10 = 5
# 5 // 10 = 0 ---> Stops here
num = n
lst = []
while num > 0:
    last_dig = num%10
    print(last_dig)
    lst.append(last_dig)
    print(lst)
    num=num//10
```

```python
# Count digits 
n = 5873
cnt = 0
num = n
num1 = n
num2 = n
lst = []
while num1 >0:
    cnt +=1
    num1//=10
print("count is : ",cnt)
# Another way 
while num2 >0:
    last_dig = num2%10 
    lst.append(last_dig)
    num2//=10
print("Another to way to count : ",len(lst))
print("List ",lst)

# Final method
from math import *
def countdig(num):
    return int(log10(num)+1)
print("3 rd way for total digits : ", countdig(num))


# TC -> O(log10(n)) , why 10 ? Because we are dividing the number by 10 to extract the last digit
# when there's a division by 5, TC -> O(log5(n))
```

```python
# Check Palindrome
def checkpalnum(n):
    num = n
    res = 0
    while num > 0:
        res = num%10 + res*10
        num //= 10
    return res == n
checkpalnum(4334)
# TC -> O(log10(n))
# SC -> O(1)
```

```python
# Armstrong Number
# 153 = 1^3 + 5^3 +3^3
# 1634 = 1^4 + 6^4 + 3^4 + 4^4
from math import *
def armstrnum(num):
    n = num
    k = int(log10(n)+1)
    lst= []
    while n > 0:
        lstdig = n%10
        lst.append(lstdig)
        n //=10
    fin = sum(i**k for i in lst)
    if fin == num :
        print("Yo Baby! That's an Armstrong number.")
    else:
        print("Better luck next time buddy.")
armstrnum(int(input()))
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Another method
##### +++++++++++++++++++++++++++++++++++++++++++ #####
def armstrongnum(num):
    n = num
    tot = 0
    k = len(str(num))
    while n > 0:
        ld = n%10
        tot = tot+(ld**k)
        n//=10
    return tot == num
armstrongnum(153)

# Both have Same  TC -> O(log10(n)+1)
# For 1st method SC -> O(log10(n)+1) as we involve a list as well
# For second SC -> O(1)
```

```python
# print factors
# Brute Force
def factor(num):
    n = num
    lst = []
    for i in range(1,n+1):
        if n%i == 0:
            lst.append(i)
    return lst
num = 45
print("Sol 1 :",factor(num))
# TC -> O(n)
# SC -> O(k) k is the number of factors stored in the list
def fac(num):
    n = num
    lst = []
    a = 1
    while a <= n:
        if n % a ==0:
            lst.append(a)
        a+=1
    return lst
print("Sol 2 :",fac(42))

##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Another method
##### +++++++++++++++++++++++++++++++++++++++++++ #####
def fact(num):
    res = []
    for i in range(1, num//2+1):
        if num%i ==0:
            res.append(i)
    res.append(num)
    return res
print("Sol 3 :", fact(10))
# TC -> O(n/2) almost similar to O(n)
# SC -> O(k)

##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Final method --> Most Optimal 
##### +++++++++++++++++++++++++++++++++++++++++++ #####
from math import sqrt
def factors(num):
    res = []
    for i in range(1, int(sqrt(num)+1)):
        if num%i==0:
            a = num//i
            res.append(i)
            if a != i:
                res.append(a)
    res.sort()
    return res
print("Sol 4 :",factors(64))
# TC -> O(sqrt(n))+O(nlogn)
# SC -> O(k)
```

```python
# Frequency Map/Dictionary
nums = [5,6,7,7,1,9,3,1,1,5,1,1,2,2,2,2]
freqmap = {} # or dict()
for i in range(0,len(nums)):
    if nums[i] in freqmap:
        freqmap[nums[i]]+=1 
    else:
        freqmap[nums[i]]=1
print(freqmap)
# We can reduce it further
freqmap1 = {} # or dict()
for i in nums:
    if i in freqmap1:
        freqmap1[i]+=1 
    else:
        freqmap1[i]=1
print(freqmap1)

# TC -> O(n)
# SC -> O(n)

##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Another method --> Using Hash Maps 
##### +++++++++++++++++++++++++++++++++++++++++++ #####
hashmap = {}
for i in nums:
    hashmap[i] = hashmap.get(i,0)+1
print(hashmap)
# TC -> O(1)
# SC -> O(1)
```
