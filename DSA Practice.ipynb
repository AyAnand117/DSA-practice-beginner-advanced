```python
# Armstrong Number
# 153 = 1^3 + 5^3 +3^3
# 1634 = 1^4 + 6^4 + 3^4 + 4^4
from math import *
def armstrnum(num):
    n = num
    k = int(log10(n)+1)
    lst= []
    while n > 0:
        lstdig = n%10
        lst.append(lstdig)
        n //=10
    fin = sum(i**k for i in lst)
    if fin == num :
        print("Yo Baby! That's an Armstrong number.")
    else:
        print("Better luck next time buddy.")
armstrnum(int(input()))
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Another method
##### +++++++++++++++++++++++++++++++++++++++++++ #####
def armstrongnum(num):
    n = num
    tot = 0
    k = len(str(num))
    while n > 0:
        ld = n%10
        tot = tot+(ld**k)
        n//=10
    return tot == num
armstrongnum(153)

# Both have Same  TC -> O(log10(n)+1)
# For 1st method SC -> O(log10(n)+1) as we involve a list as well
# For second SC -> O(1)
```

```python
# print factors
# Brute Force
def factor(num):
    n = num
    lst = []
    for i in range(1,n+1):
        if n%i == 0:
            lst.append(i)
    return lst
num = 45
print("Sol 1 :",factor(num))
# TC -> O(n)
# SC -> O(k) k is the number of factors stored in the list
def fac(num):
    n = num
    lst = []
    a = 1
    while a <= n:
        if n % a ==0:
            lst.append(a)
        a+=1
    return lst
print("Sol 2 :",fac(42))

##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Another method
##### +++++++++++++++++++++++++++++++++++++++++++ #####
def fact(num):
    res = []
    for i in range(1, num//2+1):
        if num%i ==0:
            res.append(i)
    res.append(num)
    return res
print("Sol 3 :", fact(10))
# TC -> O(n/2) almost similar to O(n)
# SC -> O(k)

##### +++++++++++++++++++++++++++++++++++++++++++ #####
# Final method --> Most Optimal 
##### +++++++++++++++++++++++++++++++++++++++++++ #####
from math import sqrt
def factors(num):
    res = []
    for i in range(1, int(sqrt(num)+1)):
        if num%i==0:
            a = num//i
            res.append(i)
            if a != i:
                res.append(a)
    res.sort()
    return res
print("Sol 4 :",factors(64))
# TC -> O(sqrt(n))+O(nlogn)
# SC -> O(k)
```

```python
# Recursion Theory
# Syntax:
#         def funct():
#                   if condition:
#                             return
#                           ---------
#                           ---------
#                   funct()

# Print 'Ayush' using recursion 4 time
cnt = 0
def functCall():
    global cnt # without this this will result in error as in the function
    if cnt==4: # python treats cnt as a local variable and not a global one
        print("-------------------------------HEAD RECURSION-------------------------------")
        return
    print("Ayush")
    cnt+=1
    functCall()
functCall()
# TC -> O(n+1) --> O(N)
# The above is a HEAD RECURSION
# SC -> O(n)
# -----------------------------------------
cnt = 0
def fCall():
    global cnt
    if cnt == 4:
        print("-------------------------------TAIL RECURSION-------------------------------")
        return
    cnt+=1
    fCall()
    print("Ayush")
fCall()
# TC -> O(n+1) --> O(N)
# SC -> O(n)
# We cna see that the function is called in tail recursion first and then the 
# job - print('Ayush') - gets done. It can also be called as BACK TRACKING.
# For every recursion problem we will have to create a recursion tree.
```

```python
# Parameterized and Funcitional Recursion
#---------------------------------------------
# Sum of 1 to N [PARAMETERIZED RECURSION]
print("----PARAMETERIZED RECURSION----")
def parrecsum(sum, i, n):
    if i > n:
        print(sum)
        return
    parrecsum(sum+i, i+1, n)
parrecsum(0,1,10)
# TC -> O(n)

print("-------------------------------")
# Sum of  1 to N [FUNCTIONAL RECURSION]
# Steps :
# - Create a flow.
# - Create a base condition
# Example : f(10) -> 10+9+8....+2+1
# it can be written as f(10) -> 10 + f(9)
# So any function's flow can be written as f(N) -> N + f(N-1)
print("----FUNCTIONAL RECURSION----")
def funrecsum(n):
    if n==1:
        return 1
    return n + funrecsum(n-1) 
funrecsum(10)
# TC -> O(n)
# SC -> O(n) - it is a stack space as the stacks gets filled up 
# but it gets cleared only after the function returns something.
```

```python
# Check if a string is palindrome
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 1st method --> Using slicing
##### +++++++++++++++++++++++++++++++++++++++++++ #####
text = input("Enter your text : ")
text = text.lower().replace(" ","")
print("----Slicing Method----")
if text==text[::-1]:
    print("text is palindrome")
else:
    print("text is not a palindrome")
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 2nd method --> Using while loop
##### +++++++++++++++++++++++++++++++++++++++++++ #####
n = len(text)
a = ""
print("----While Loop Method----")
while n>0:
    a += text[n-1]
    n-=1
if text==a:
    print("text is palindrome")
else:
    print("text is not a palindrome")
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 3rd method --> Using Two Pointers
##### +++++++++++++++++++++++++++++++++++++++++++ #####
print("----Two Pointers Method----")
b = list(text)
l = 0
r = len(b)-1
while l<r :
    b[l],b[r]=b[r],b[l]
    l+=1
    r-=1
revb = "".join(b)
if text == revb:
    print("text is palindrome")
else:
    print("text is not a palindrome")
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 4th method --> Two pointers with comparison
##### +++++++++++++++++++++++++++++++++++++++++++ #####
print("----Two Pointers with Comparison Method----")
n = len(text)
l =0 
r = n-1
is_pal = True
while l<r:
    if text[l]!=text[r]:
        is_pal = False
        break
    l+=1
    r-=1
if is_pal:
    print("text is palindrome")
else:
    print("text is not a palindrome")
# TC -> O(n/2) -> similar to O(N)
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 5th method --> Using Recursion
##### +++++++++++++++++++++++++++++++++++++++++++ #####
print("----Recursion Method----")
n = len(text)
l = 0
r = n-1
def palindrome(text, l, r):
    var = False
    if l>=r:
        var = True
        return var
    if text[l]!=text[r]:
        var = False
        return var
    return palindrome(text,l+1, r-1)
res = palindrome(text,l,r)
if res:
    print("text is palindrome")
else:
    print("text is not a palindrome")
```

```python
# Selection Sort
# Ascending order
nums = [5,7,8,4,1,6,9,2,3]
print("Array : ",nums)
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 1st method --> Using Brute Force
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# length = 9
# Take first element as the ith index and consider it as the minimum index
print("----Brute Force Method----")
def selectsort(nums):
    n = len(nums)
    for i in range(0,n):
        minidx = i
        for j in range(i+1,n):
            if nums[j]<nums[minidx]:
                minidx = j
        nums[i],nums[minidx]=nums[minidx],nums[i]
    print("Sorted array : ",nums)
selectsort(nums)
# TC --> O(n^2) 
# SC --> O(1)
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 2nd method --> Using Bubble Sort
##### +++++++++++++++++++++++++++++++++++++++++++ #####
print("----Bubble Sort----")
# Bubble sort works on adjacent swap
def bubblesort(nums):
    lst = nums
    n = len(nums)
    for i in range(n-2,-1,-1):
        #is_swap = False
        for j in range(0,i+1):
            if nums[j]>nums[j+1]:
                nums[j],nums[j+1]=nums[j+1],nums[j]
       #         is_swap = True
      #  if is_swap == False:
      #      print("SOrted list")
    print(f"Bubble Sorted : {lst}")
bubblesort(nums)
# TC --> O(n^2)
# SC --> O(1)
# This is average and worst case
##### +++++++++++++++++++++++++++++++++++++++++++ #####
# 3rd method --> Insertion Sort
##### +++++++++++++++++++++++++++++++++++++++++++ #####
print("----Insertion Sort----")
def insertionsort(nums):
    n = len(nums)
    for i in range(1,n):
        k = nums[i]
        j = i-1
        while j >= 0 and nums[j]>k:
            nums[j+1] = nums[j]
            j-=1
    print(nums)
insertionsort(nums)
```
