```python
# Rough Code
nums3 = [5,7,8,4,1,6,9,2,3]  
n = len(nums3)  
left = nums3[0:n//2]  
right = nums3[n//2:n]  
res = []  
for i,j in zip(left,right):  
    if i>=j:  
        res.append(j)  
        res.append(i)  
    else:  
        res.append(i)  
        res.append(j)  
print(res)  

# Before performing mergesort(nums)
# Lets try merging two sorted list first
arr1 = [1,2,3,4,5]   
arr2 = [1,6,7,8]  
def merge_arr(l,r):  
    res=[]  
    i,j=0,0  
    n,m=len(l),len(r)  
    while i<n and j<m:  
        if l[i]<=r[j]:  
            res.append(l[i])  
            i+=1  
        else:  
            res.append(r[j])  
            j+=1  
    if i<n:  
        while i<n:  
            res.append(l[i])  
            i+=1  
    if j<m:  
        while j<m:  
            res.append(r[j])  
            j+=1  
    return res  
merge_arr(arr1,arr2)  
```
