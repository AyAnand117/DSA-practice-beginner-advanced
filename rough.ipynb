```python
def rearrangeSign(arr):
    # Separate into positives and negatives
    pos = [x for x in arr if x >= 0]
    neg = [x for x in arr if x < 0]
    
    res = []
    i = 0
    
    # Interleave them as long as both lists have elements
    while i < len(pos) and i < len(neg):
        res.append(pos[i])
        res.append(neg[i])
        i += 1
        
    # Append any remaining leftover elements from either list
    res.extend(pos[i:])
    res.extend(neg[i:])
    
    return res
```
